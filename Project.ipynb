```python
import requests
import pandas as pd

url = "https://pokeapi.co/api/v2/pokemon?limit=100000"

response = requests.get(url)
data = response.json()

data

pokemon_list = {pokemon['name'] : pokemon['url'] for pokemon in data['results']}

def get_pokemon_details(pokemon_url):
    response = requests.get(pokemon_url)
    pokemon_data = response.json()
    
    name = pokemon_data["name"]
    id_ = pokemon_data["id"]
    height = pokemon_data["height"]
    weight = pokemon_data["weight"]
    types = [t["type"]["name"] for t in pokemon_data["types"]]
    abilities = [a["ability"]["name"] for a in pokemon_data["abilities"]]
    base_stats = {s["stat"]["name"]: s["base_stat"] for s in pokemon_data["stats"]}
    front_pic_link = pokemon_data['sprites']['front_default']
    back_pic_link = pokemon_data['sprites']['back_default']
    
    return {
        "id": id_,
        "name": name,
        "height": height,
        "weight": weight,
        "types": types,
        "abilities": abilities,
        "base_stats": base_stats,
        "front_pic_link": front_pic_link,
        "back_pic_link": back_pic_link
    }

compilation = pd.DataFrame(columns = ['id','name','height','weight','types','abilities','base_stats', 'front_pic_link', 'back_pic_link'])
print(len(pokemon_list))
count = 0
for pokemon, pokemon_url in pokemon_list.items():
    output = get_pokemon_details(pokemon_url)
    outputData = list(output.values())
    compilation.loc[len(compilation)] = outputData
    count += 1
    print(f'completed: {count}/{len(pokemon_list)}')
    
compilation.to_csv('PokeAPI_output.csv', index=False)
```
